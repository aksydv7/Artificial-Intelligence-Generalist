```python
""" Core Data Types [33:46-44:19]:
Integer (int)
Float (float)
String (str) (including the use of single, double, and triple quotes, and string replication with multiplication)
Complex (complex) (represented with j for the imaginary part)
Boolean (bool) (True/False, which Python internally treats as 1/0 for arithmetic operations)
"""
# python is dynamically typed, meaning you don't have to declare the type of a variable explicitly. Python will infer the type based on the value assigned to it.

x = 90
type(x)  # This will return <class 'int'>
type(3.14)  # This will return <class 'float'>
type("Hello")  # This will return <class 'str'>
type(2 + 3j)  # This will return <class 'complex'>
type(True)  # This will return <class 'bool'>
type(False)  # This will return <class 'bool'>

# type() function is used to check the type of a variable or value in Python. It returns the class type of the object passed to it.

```

```python
# type conversion functions in Python are used to convert a value from one data type to another. Here are some commonly used type conversion functions:
# int(): Converts a value to an integer.
# float(): Converts a value to a floating-point number.
# str(): Converts a value to a string.
# bool(): Converts a value to a boolean.

# there are also other type conversion functions like list(), tuple(), set(), dict() etc. which are used to convert values to their respective data types.
# implisit type conversion (also known as type coercion) is when Python automatically converts one data type to another during an operation. For example, if you add an integer and a float, Python will implicitly convert the integer to a float before performing the addition.
# explicit type conversion (also known as type casting) is when you manually convert a value from one data type to another using the type conversion functions mentioned above.

x = 5  # integer
y = 2.5  # float
z = x + y  # implicit type conversion, x is converted to float
print(z)  # Output: 7.5

print(int(3.14))  # explicit type conversion, converts float to int, Output: 3
print(float(5))   # explicit type conversion, converts int to float, Output: 5.0
print(str(123))   # explicit type conversion, converts int to string, Output: "123"
print(bool(0))    # explicit type conversion, converts int to boolean, Output: False
print(bool(1))    # explicit type conversion, converts int to boolean, Output: True
print(bool("hello"))  # explicit type conversion, converts string to boolean, Output: True
print(bool(""))     # explicit type conversion, converts empty string to boolean, Output: False


# Truthy: Any string that contains at least one character is considered True.
# Falsy: An empty string (a string with absolutely zero characters) is considered False
print("truthy string" == True)  # Output: False, because a non-empty string is truthy, but not equal to the boolean value True
print(bool("truthy string"))  # Output: True, because a non-empty string is
bool(0)
bool(1)
print("Is this truthy?", bool(-0.1))  # Output: True, because any non-zero number is considered truthy

# other truthy values include non-empty lists, tuples, sets, and dictionaries. For example:
print(bool([1, 2, 3]))  # Output: True, because the list is non-empty
print(bool((1, 2)))    # Output: True, because the tuple is non-empty
print(bool({1, 2, 3}))  # Output: True, because the set is non-empty
print(bool({"key": "value"}))  # Output: True, because the dictionary is non-empty     
print(bool([]))  # Output: False, because the list is empty
print(bool(()))  # Output: False, because the tuple is empty   
print(bool({}))  # Output: False, because the dictionary is empty
```

```python
# print() Function [1:13:21-1:16:59]: Its purpose for displaying output to the console and the concept of its NoneType return value, which prevents arithmetic operations directly on the function call.
print("Hello, World!")  # Output: Hello, World!
print(type(print("Hello, World!")))  # Output: <class 'NoneType'>
print(print())  # Output: None

# We assign the result of the print function to the variable 'x'
x = print("Hello World!")

# Let's check what 'x' actually holds
print(x)
print(type(x))
```
