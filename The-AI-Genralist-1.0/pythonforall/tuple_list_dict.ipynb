```python
# list is a mutable data structure, which means you can change its content after it has been created. You can add, remove, or modify elements in a list using various methods and operations.

list01 = ["Akash", "Syd", "Vishal", "Rohit"]
list01[1] = "Romit"
list01 
```

```python
# Indexing: Accessing Elements
print(list01[0])  # Output: Akash
print(list01[1])  # Output: Romit
print(list01[2])  # Output: Vishal
print(list01[3])  # Output: Rohit

print(list01[-1])  # Output: Rohit (last element)
print(list01[-2])  # Output: Vishal (second to last element)
print(list01[-3])  # Output: Syd (third to last element)
print(list01[-4])  # Output: Akash (fourth to last element)

len(list01)  # Output: 4 (number of elements in the list)
# if we want a list elements in reverse order we can use slicing with a step of -1
count_rev = -len(list01)
print(count_rev)

```

```python
# ['Akash', 'Romit', 'Vishal', 'Rohit']

count_rev = -len(list01)
print(count_rev)
print("before loop count rev:", count_rev)

while count_rev < 0:
    print(list01[count_rev], count_rev)
    count_rev += 1
    
print("after loop count rev:", count_rev)
```

```python
# Positive indexing: Starts at 0 (first element) and goes up to n-1 (last element).

# Negative indexing: Starts at -1 (last element) and goes down to -n (first element).

alphabet = ["A", "B", "C", "D", "E", "F", "G"]

length_alphabet = len(alphabet)
# 1. Basic Slicing (Start at index 1, stop BEFORE index 4)
print(alphabet[1:4]) 
# Output: ['B', 'C', 'D']

# 2. Omitting Start or End
print(alphabet[:3])  # Starts from the beginning (index 0) up to index 3
# Output: ['A', 'B', 'C']

print(alphabet[4:])  # Starts at index 4 and goes to the very end
# Output: ['E', 'F', 'G']

# 3. Using the Step argument (Every 2nd item)
print(alphabet[0:7:2]) 
# Output: ['A', 'C', 'E', 'G']

# 4. 💡 Pro-Tip: Reversing a list using negative step
print(alphabet[::-1]) 
# Output: ['G', 'F', 'E', 'D', 'C', 'B', 'A']
```

```python
colors = ["Red", "Blue", "Green", "Blue", "Yellow"]

# list methods for adding elements

colors.append("Purple")  # Adds "Purple" to the end of the list
print(colors)  # Output: ['Red', 'Blue', 'Green', 'Blue', 'Yellow', 'Purple']

colors.insert(2, "Orange")  # Inserts "Orange" at index 2
print(colors)  # Output: ['Red', 'Blue', 'Orange', 'Green', 'Blue', 'Yellow', 'Purple']

extra_colors = ["Pink", "Brown"]
colors.extend(extra_colors)  # Extends the list by adding elements from extra_colors
print(colors)  # Output: ['Red', 'Blue', 'Orange', 'Green', 'Blue', 'Yellow', 'Purple', 'Pink', 'Brown']

```

```python
# list methods for removing elements

colors.remove("Blue")  # Removes the first occurrence of "Blue"
print(colors)  # Output: ['Red', 'Orange', 'Green', 'Blue', 'Yellow', 'Purple', 'Pink', 'Brown']

print(colors.pop())
print(colors)  # Output: ['Red', 'Orange', 'Green', 'Blue', 'Yellow', 'Purple', 'Pink'] 

colors.pop(0)
colors 

# colors.clear()  # Removes all elements from the list
print(colors)  # Output: []
```

```python
# Serching & analysing elements in a list

colors.reverse()
colors

colors.sort()  # Sorts the list in ascending order
colors

colors.index("Green")  # Returns the index of the first occurrence of "Green"
# colors.index("Akash")
colors.count("")
colors.count("Blue")  # Returns the number of occurrences of "Blue" in the list

len(colors) # Returns the number of elements in the list
```

```python
#  The Core Rule: Immutability
# "Immutable" simply means "unchangeable." Once a tuple is created in computer memory, you cannot add items, remove items, or alter existing items.

tuple_example = (1, 2, 3, 4, 5)
print(tuple_example)

tuple_example[0] = 10  # This will raise a TypeError because tuples are immutable
```

```python
# slicing rules will apply on tuples as well as lists. The slicing rules are the same for both data structures. You can use the same syntax to access elements in a tuple as you would in a list.

rainbow = ("Red", "Orange", "Yellow", "Green", "Blue", "Indigo", "Violet")

# Indexing
print(rainbow[0])    # Output: Red
print(rainbow[-1])   # Output: Violet

# Slicing (Grabbing Orange, Yellow, Green)
# Start at index 1, stop BEFORE index 4
print(rainbow[1:4])  # Output: ('Orange', 'Yellow', 'Green')

print(rainbow[:3])   # Output: ('Red', 'Orange', 'Yellow')
print(rainbow[4:])   # Output: ('Blue', 'Indigo', 'Violet') 
print(rainbow[::2])  # Output: ('Red', 'Yellow', 'Blue', 'Violet')
```
